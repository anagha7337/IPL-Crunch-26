```python
def get_phase_diff_scores(df, value_col):
    scores = {
        "Powerplay": 0,
        "Middle Overs": 0,
        "Death Overs": 0
    }

    for match_id in df["match_id"].unique():

        match_df = df[
            df["match_id"] == match_id
        ]

        phases = {}

        for phase in [
            "Powerplay",
            "Middle Overs",
            "Death Overs"
        ]:

            phase_df = match_df[
                match_df["phase"] == phase
            ]

            if len(phase_df) < 2:
                continue

            values = phase_df[value_col].tolist()

            diff = abs(values[0] - values[1])

            phases[phase] = diff

        if phases:
            biggest = max(
                phases,
                key=phases.get
            )

            scores[biggest] += 1

    return scores
```

```python
run_scores = get_phase_diff_scores(
    phase_runs,
    "runs_total"
)

print(run_scores)
```

```python
wicket_scores = get_phase_diff_scores(
    phase_wickets,
    "is_wicket"
)

print(wicket_scores)
```

```python
matches["batting_first"] = matches.apply(
    lambda x: x["toss_winner"]
    if x["toss_decision"] == "bat"
    else (
        x["team2"]
        if x["toss_winner"] == x["team1"]
        else x["team1"]
    ),
    axis=1,
)

matches["bowling_first"] = matches.apply(
    lambda x: x["team2"]
    if x["batting_first"] == x["team1"]
    else x["team1"],
    axis=1,
)

tw_bat = matches[
    (matches["toss_winner"] == matches["batting_first"]) &
    (matches["winner"] == matches["toss_winner"])
].shape[0]

tw_bowl = matches[
    (matches["toss_winner"] == matches["bowling_first"]) &
    (matches["winner"] == matches["toss_winner"])
].shape[0]

tl_bat = matches[
    (matches["toss_winner"] != matches["batting_first"]) &
    (matches["winner"] == matches["batting_first"])
].shape[0]

tl_bowl = matches[
    (matches["toss_winner"] != matches["bowling_first"]) &
    (matches["winner"] == matches["bowling_first"])
].shape[0]

print("Toss win + batting first :", tw_bat)
print("Toss win + bowling first :", tw_bowl)
print("Toss lose + batting first:", tl_bat)
print("Toss lose + bowling first:", tl_bowl)
```

```python
results = pd.Series({
    "Toss Win + Bat First": tw_bat,
    "Toss Win + Bowl First": tw_bowl,
    "Toss Lose + Bat First": tl_bat,
    "Toss Lose + Bowl First": tl_bowl
})

results.plot(kind="bar", figsize=(8,5))
plt.ylabel("Wins")
plt.title("IPL Wins by (Toss + Toss Decision)")
plt.tight_layout()
plt.savefig("../charts/toss+toss_decision.png")
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt

def phase(over):
    if over <= 5:
        return "Powerplay"
    elif over <= 15:
        return "Middle Overs"
    else:
        return "Death Overs"

matches["phase"] = matches["over"].apply(phase)

matches["result"] = matches.apply(
    lambda x: "Winning Teams"
    if x["batting_team"] == x["winner"]
    else "Losing Teams",
    axis=1
)

phase_runs = (
    matches.groupby(
        ["match_id", "innings", "batting_team", "phase", "result"]
    )["runs_total"]
    .sum()
    .reset_index()
)

avg_runs = (
    phase_runs.groupby(["phase", "result"])["runs_total"]
    .mean()
    .unstack()
)

avg_runs = avg_runs.loc[
    ["Powerplay", "Middle Overs", "Death Overs"]
]

avg_runs.plot(
    kind="bar",
    figsize=(8,5)
)

plt.title("Average Runs per Phase: Winning vs Losing Teams")
plt.xlabel("Phase")
plt.ylabel("Average Runs")
plt.xticks(rotation=0)

plt.tight_layout()
plt.savefig("../average_runs_per_phase.png")
plt.show()
```

```python
import pandas as pd

df = matches.sort_values(
    ["match_id", "innings", "over", "ball"]
).copy()

df["is_noball"] = df["extras_noballs"] > 0

df["free_hit"] = (
    df.groupby(["match_id", "innings"])["is_noball"]
    .shift(1)
    .fillna(False)
)
free_hits = df[df["free_hit"]].copy()

free_hits["boundary"] = free_hits["runs_batter"] >= 4

strike_rate = (
    free_hits["runs_batter"].sum()
    / len(free_hits)
) * 100

boundary_pct = (
    free_hits["boundary"].mean()
) * 100

dot_pct = (
    (free_hits["runs_batter"] == 0).mean()
) * 100

avg_runs = free_hits["runs_batter"].mean()

print("Free Hit deliveries:", len(free_hits))
print("Strike Rate:", round(strike_rate, 2))
print("Boundary %:", round(boundary_pct, 2))
print("Dot Ball %:", round(dot_pct, 2))
print("Avg Runs:", round(avg_runs, 2))
```

```python
import matplotlib.pyplot as plt
import pandas as pd

metrics = pd.Series({
    "Boundary %": boundary_pct,
    "Dot Ball %": dot_pct,
    "Strike Rate": strike_rate
})

ax = metrics.plot(
    kind="bar",
    figsize=(7,5)
)

plt.title("Free Hit Outcomes in IPL")
plt.ylabel("Percentage / Strike Rate")
plt.xticks(rotation=0)

for i, v in enumerate(metrics):
    ax.text(
        i,
        v + 2,
        f"{v:.1f}",
        ha="center"
    )

plt.tight_layout()
plt.savefig("../charts/free_hit_outcomes.png")
plt.show()
```

```python
import matplotlib.pyplot as plt

innings = df.groupby(
    ["match_id", "batting_team", "batter"],
    as_index=False
).agg(
    runs=("runs_batter", "sum"),
    balls=("ball", "count")
)

innings["strike_rate"] = (
    innings["runs"] / innings["balls"]
) * 100

winners = (
    df.groupby("match_id")["winner"]
    .first()
    .reset_index()
)

innings = innings.merge(winners, on="match_id")

slow_anchor = innings[
    (innings["runs"] >= 50) &
    (innings["strike_rate"] < 130)
].copy()

quick_cameo = innings[
    (innings["runs"] >= 25) & (innings["runs"] <= 50) &
    (innings["strike_rate"] > 200)
].copy()

slow_anchor["won"] = (
    slow_anchor["batting_team"] == slow_anchor["winner"]
)

quick_cameo["won"] = (
    quick_cameo["batting_team"] == quick_cameo["winner"]
)

labels = [
    "Slow Anchor\n50+ SR<130",
    "Quick Cameo\n25 <= Runs <= 50 SR>200"
]

win_rates = [
    slow_anchor["won"].mean() * 100,
    quick_cameo["won"].mean() * 100
]

plt.figure(figsize=(7,5))

bars = plt.bar(labels, win_rates)

plt.ylabel("Team Win Percentage")
plt.title("IPL: Slow Anchors vs Quick Cameos")
plt.ylim(0,100)


for bar in bars:
    y = bar.get_height()
    plt.text(
        bar.get_x() + bar.get_width()/2,
        y + 2,
        f"{y:.1f}%",
        ha="center"
    )

plt.savefig("../charts/slow_anchor_vs_quick_cameo(1).png")
plt.show()

print("Slow anchor innings:", len(slow_anchor))
print("Quick cameo innings:", len(quick_cameo))
```
